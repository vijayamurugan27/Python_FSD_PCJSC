# Variables - Primitive Data Types

In Python, variables are dynamically typed containers used to store data values in memory.
Python has five primary primitive (scalar) data types:
- **String (`str`)**
- **Integer (`int`)**
- **Float (`float`)**
- **Complex (`complex`)**
- **Boolean (`bool`)**

## 1. String (`str`)

### Program Syntax
```python
variable_name = "value"  # or 'value' or """multiline"""
```

### Why it is being used
Strings are used to store and manipulate textual information such as names, messages, addresses, and user input.

````python
# Sample code: String
message = "Hello, Python!"

print("Value:", message)
print("Data Type:", type(message))
````

### Explanation
1. The variable `message` stores a sequence of characters enclosed within double quotes.
2. The `print()` function displays the text, and `type()` confirms the variable belongs to the `str` class.
3. Strings in Python are immutable, meaning their individual characters cannot be modified in-place.

---
## 2. Integer (`int`)

### Program Syntax
```python
variable_name = whole_number  # positive, negative, or zero
```

### Why it is being used
Integers are used for counting items, indexing collections, tracking loop iterations, and performing exact arithmetic without decimals.

````python
# Sample code: Integer
student_count = 45

print("Value:", student_count)
print("Data Type:", type(student_count))
````

### Explanation
1. The variable `student_count` stores a whole numerical value without any decimal point.
2. Calling `type(student_count)` verifies that the object belongs to Python's `int` class.
3. Python integers support arbitrary precision, meaning they can grow as large as memory permits without integer overflow.

---
## 3. Float (`float`)

### Program Syntax
```python
variable_name = decimal_number  # e.g., 19.99 or scientific notation 1.5e3
```

### Why it is being used
Floats are used when representing fractional quantities, continuous measurements, monetary values, and scientific calculations.

````python
# Sample code: Float
item_price = 49.99

print("Value:", item_price)
print("Data Type:", type(item_price))
````

### Explanation
1. The variable `item_price` is assigned a real number containing a decimal component.
2. The `print()` function outputs the numeric value, and `type()` identifies it as a `float`.
3. Floats in Python follow the IEEE 754 standard for double-precision (64-bit) floating-point numbers.

---
## 4. Complex (`complex`)

### Program Syntax
```python
variable_name = real_part + imag_part * 1j  # or complex(real, imag)
```

### Why it is being used
Complex numbers are used in mathematical computing, signal processing, electrical engineering (AC circuits), and scientific simulations.

````python
# Sample code: Complex
impedance = 3 + 4j

print("Value:", impedance)
print("Real Part:", impedance.real)
print("Imaginary Part:", impedance.imag)
print("Data Type:", type(impedance))
````

### Explanation
1. The variable `impedance` contains a complex number with real component `3` and imaginary component `4j`.
2. Built-in attributes `.real` and `.imag` access the components individually as floating-point numbers.
3. Python supports complex arithmetic natively without needing external mathematical libraries.

---
## 5. Boolean (`bool`)

### Program Syntax
```python
variable_name = True   # or False (capitalized)
```

### Why it is being used
Booleans are used to control decision-making logic (`if`/`else`), loop conditions, validation checks, and binary flags.

````python
# Sample code: Boolean
is_course_active = True

print("Value:", is_course_active)
print("Data Type:", type(is_course_active))
````

### Explanation
1. The variable `is_course_active` is assigned the boolean literal `True`.
2. The `type()` function verifies that the variable is an instance of the `bool` class.
3. In Python, `bool` is a subclass of `int`, where `True` evaluates to 1 and `False` evaluates to 0 in numeric operations.

# Non-Primitive Data Types (Collections & Containers)

Non-primitive data types store collections of values rather than a single scalar value.
The primary built-in collection and container types in Python are:
- **List (`list`)**
- **Tuple (`tuple`)**
- **Set (`set`)**
- **Dictionary / Container (`dict`)**

## 1. List (`list`)

### 1) What is
An ordered, mutable collection of elements that allows duplicate values and can store heterogeneous data types.

### 2) Creation
```python
empty_list = []  # or list()
fruits = ["apple", "banana", "cherry"]
```

### 3) USP (Unique Selling Proposition)
Full mutability with ordered sequence indexing, allowing dynamic additions, deletions, and in-place reordering.

````python
# Creation
fruits = ["apple", "banana", "cherry"]

# Inbuilt Methods
fruits.append("orange")     # Adds to end
fruits.insert(1, "mango")   # Inserts at index 1
removed = fruits.pop()      # Removes and returns last item
fruits.sort()               # Sorts list in-place

print("Updated List:", fruits)
print("Removed Element:", removed)
````

### Explanation
1. Lists maintain an ordered collection of changeable items accessible by zero-based index numbers.
2. Built-in methods like `append()`, `insert()`, `pop()`, and `sort()` allow fast dynamic modification of elements.

---
## 2. Tuple (`tuple`)

### 1) What is
An ordered, immutable collection of elements that allows duplicate values and cannot be modified once defined.

### 2) Creation
```python
empty_tuple = ()  # or tuple()
coordinates = (10.0, 20.0, 10.0)  # (val,) with comma for single item
```

### 3) USP (Unique Selling Proposition)
Guaranteed immutability ensuring data integrity, memory efficiency, faster execution, and eligibility to serve as dictionary keys.

````python
# Creation
coordinates = (10.0, 20.0, 10.0)

# Inbuilt Methods
freq = coordinates.count(10.0)  # Counts occurrences of value
idx = coordinates.index(20.0)   # Finds index of first occurrence

print("Tuple:", coordinates)
print("Count of 10.0:", freq)
print("Index of 20.0:", idx)
````

### Explanation
1. Tuples provide a secure, read-only sequence whose elements cannot be added, changed, or deleted after creation.
2. They feature lightweight inspection methods such as `count()` and `index()` without any modifying operations.

---
## 3. Set (`set`)

### 1) What is
An unordered collection of unique elements that automatically discards duplicate entries and supports mathematical operations.

### 2) Creation
```python
empty_set = set()  # Note: {} creates an empty dict, not a set
colors = {"red", "green", "blue"}
```

### 3) USP (Unique Selling Proposition)
Strict element uniqueness with O(1) constant-time membership lookups and native mathematical set algebra (union, intersection).

````python
# Creation (duplicate 'red' is automatically dropped)
colors = {"red", "green", "blue", "red"}

# Inbuilt Methods
colors.add("yellow")             # Adds single element
colors.discard("green")          # Safely removes element
others = {"blue", "purple"}
common = colors.intersection(others) # Finds common elements

print("Set Elements:", colors)
print("Intersection with others:", common)
````

### Explanation
1. Sets automatically filter out duplicate values and store elements in an unindexed, hash-based structure.
2. Methods like `add()`, `discard()`, and `intersection()` provide fast item management and mathematical relations.

---
## 4. Container / Dictionary (`dict`)

### 1) What is
An associative key-value mapping container where each unique immutable key maps to an arbitrary data value.

### 2) Creation
```python
empty_dict = {}  # or dict()
student = {"id": 101, "name": "Alice", "grade": "A"}
```

### 3) USP (Unique Selling Proposition)
Ultra-fast O(1) average lookup, insertion, and key-based retrieval via hash table indexing rather than sequential scanning.

````python
# Creation
student = {"id": 101, "name": "Alice", "grade": "A"}

# Inbuilt Methods
student_name = student.get("name")                    # Safe lookup
student.update({"grade": "A+", "city": "New York"})   # Updates/adds keys
keys_list = list(student.keys())                      # Retrieves all keys
values_list = list(student.values())                  # Retrieves all values

print("Updated Student Dict:", student)
print("Retrieved Name:", student_name)
print("Keys:", keys_list)
````

### Explanation
1. Dictionaries organize data into labeled key-value pairs enabling direct associative access by key name.
2. Methods such as `get()`, `update()`, and `keys()` provide safe querying and dynamic modifications.

# Flow Control Statements

Flow control statements direct the order of execution in a Python program based on boolean conditions.
Conditional branching statements covered:
- **`if` Statement**
- **`if-else` Statement**
- **`elif` Statement**
- **`nested-if` Statement**

## 1. `if` Statement

### Syntax
```python
if condition:
    # statement(s) executed when condition is True
```

````python
# Simple if statement
temperature = 32

if temperature > 30:
    print("It's a hot day!")
````

### Explanation
1. The `if` statement evaluates whether the condition `temperature > 30` is `True`.
2. Because the condition is met, the indented print statement executes.

---
## 2. `if-else` Statement

### Syntax
```python
if condition:
    # statement(s) executed when condition is True
else:
    # statement(s) executed when condition is False
```

````python
# Simple if-else statement
age = 16

if age >= 18:
    print("Eligible to vote.")
else:
    print("Not eligible to vote yet.")
````

### Explanation
1. The `if` condition tests if `age >= 18`, which evaluates to `False`.
2. Execution automatically diverts to the `else` branch to run the alternate code block.

---
## 3. `elif` Statement

### Syntax
```python
if condition_1:
    # statement(s) for condition_1
elif condition_2:
    # statement(s) for condition_2
else:
    # statement(s) when no condition is satisfied
```

````python
# Simple elif ladder
marks = 85

if marks >= 90:
    print("Grade: A")
elif marks >= 80:
    print("Grade: B")
else:
    print("Grade: C")
````

### Explanation
1. The `elif` clause checks subsequent conditions sequentially once earlier conditions evaluate to `False`.
2. As soon as a condition matches (`marks >= 80`), its block runs and the remaining chain is skipped.

---
## 4. `nested-if` Statement

### Syntax
```python
if outer_condition:
    if inner_condition:
        # statement(s) executed when both conditions are True
```

````python
# Simple nested-if statement
has_ticket = True
has_id_proof = True

if has_ticket:
    if has_id_proof:
        print("Entry allowed: Ticket and ID verified.")
    else:
        print("Entry denied: Missing ID proof.")
````

### Explanation
1. A nested `if` statement places one conditional decision block inside another.
2. The inner condition is evaluated only after the outer condition evaluates to `True`.

# Loops (Iterative Statements)

Loops are used to repeatedly execute a block of code either over a sequence or while a condition remains True.
Loop types covered:
- **`for` loop using `range()`**
- **`for` loop over a `list`**
- **`while` loop**

## 1. `for` Loop using `range()`

### Syntax
```python
for variable in range(start, stop, step):
    # statement(s) executed per iteration
```

````python
# For loop using range()
for count in range(1, 6):
    print("Count:", count)
````

### Explanation
1. The `range(1, 6)` generates an arithmetic sequence from 1 up to (but excluding) 6.
2. The loop variable `count` takes each number sequentially, running the code block 5 times.

---
## 2. `for` Loop over a `list`

### Syntax
```python
for item in list_name:
    # statement(s) executed per item
```

````python
# For loop over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print("Fruit:", fruit)
````

### Explanation
1. The `for` loop directly traverses through the `fruits` list item by item without index offsets.
2. In each iteration, `fruit` receives the next list element until all items are processed.

---
## 3. `while` Loop

### Syntax
```python
while condition:
    # statement(s) executed while condition remains True
    # condition update / increment
```

````python
# While loop
num = 1

while num <= 3:
    print("Iteration:", num)
    num += 1
````

### Explanation
1. The `while` loop continuously executes as long as the condition `num <= 3` evaluates to `True`.
2. The counter `num` increments by 1 each cycle, eventually making the condition `False` to exit safely.

# Operators in Python

Operators are special symbols used to perform computations on variables and values.
Categories covered:
- **Arithmetic Operators**
- **Assignment Operators**
- **Comparison Operators**
- **Logical Operators**
- **Identity Operators**
- **Membership Operators**
- **Bitwise Operators**

## 1. Arithmetic Operators

### Syntax
```python
result = a + b  # +, -, *, /, //, %, **
```

````python
# Arithmetic operators
a, b = 10, 3

print("Addition (+):", a + b)
print("Division (/):", a / b)
print("Floor Division (//):", a // b)
print("Modulus (%):", a % b)
print("Exponentiation (**):", a ** b)
````

### Explanation
1. Arithmetic operators perform common mathematical computations between numeric values.
2. Python supports standard arithmetic like `+` and `/`, plus floor division `//`, modulus `%`, and powers `**`.

---
## 2. Assignment Operators

### Syntax
```python
variable = value   # Simple assignment
variable += value  # Compound assignment (e.g. +=, -=, *=, /=)
```

````python
# Assignment operators
x = 10
print("Initial x:", x)

x += 5   # Equivalent to x = x + 5
print("After x += 5:", x)

x *= 2   # Equivalent to x = x * 2
print("After x *= 2:", x)
````

### Explanation
1. Assignment operators store the right-hand value into the left-hand variable.
2. Compound operators like `+=` and `*=` compute an operation and reassign the result in a single step.

---
## 3. Comparison Operators

### Syntax
```python
result = a == b  # ==, !=, >, <, >=, <=
```

````python
# Comparison operators
p, q = 20, 15

print("Equal (==):", p == q)
print("Not Equal (!=):", p != q)
print("Greater than (>):", p > q)
print("Less than or equal (<=):", p <= q)
````

### Explanation
1. Comparison operators compare two operands and evaluate to a boolean `True` or `False`.
2. They are fundamental in constructing conditions for `if` statements and loops.

---
## 4. Logical Operators

### Syntax
```python
result = condition1 and condition2  # and, or, not
```

````python
# Logical operators
has_license = True
has_insurance = False

print("AND (both True):", has_license and has_insurance)
print("OR (at least one True):", has_license or has_insurance)
print("NOT (inverts boolean):", not has_license)
````

### Explanation
1. Logical operators combine or negate boolean conditions to make compound decisions.
2. `and` returns `True` only if both sides are true, `or` requires one true side, and `not` flips the truth value.

---
## 5. Identity Operators

### Syntax
```python
result = a is b      # is, is not
```

````python
# Identity operators
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("list1 == list2 (same values):", list1 == list2)
print("list1 is list2 (same object in memory):", list1 is list2)
print("list1 is list3 (same object in memory):", list1 is list3)
print("list1 is not list2:", list1 is not list2)
````

### Explanation
1. Identity operators (`is`, `is not`) check whether two variables share the exact same memory address (`id()`).
2. Unlike `==` which checks value equality, `is` verifies whether both variables reference the identical object.

---
## 6. Membership Operators

### Syntax
```python
result = item in sequence  # in, not in
```

````python
# Membership operators
fruits = ["apple", "banana", "cherry"]

print("'apple' in fruits:", "apple" in fruits)
print("'orange' in fruits:", "orange" in fruits)
print("'orange' not in fruits:", "orange" not in fruits)
````

### Explanation
1. Membership operators test whether a target element exists within a sequence like a list, string, or set.
2. `in` returns `True` if found, while `not in` returns `True` when the item is absent.

---
## 7. Bitwise Operators

### Syntax
```python
result = a & b  # &, |, ^, ~, <<, >>
```

````python
# Bitwise operators (operating on binary representation)
# a = 5 (0101 in binary), b = 3 (0011 in binary)
a, b = 5, 3

print("Bitwise AND (a & b):", a & b)    # 0001 -> 1
print("Bitwise OR (a | b):", a | b)     # 0111 -> 7
print("Bitwise XOR (a ^ b):", a ^ b)    # 0110 -> 6
print("Bitwise NOT (~a):", ~a)          # -(a+1) -> -6
print("Left Shift (a << 1):", a << 1)   # 1010 -> 10
````

### Explanation
1. Bitwise operators perform direct binary bit-by-bit manipulation on integer values.
2. They are used in low-level systems programming, network masking, and fast bit-level flags.

# Functions & Comprehensions in Python

Functions are organized, reusable blocks of code used to perform a single related action.
Comprehensions provide concise and expressive syntax for constructing sequences and mappings.
Topics covered:
- **Function without Return Value (Void Function)**
- **Function with Return Value**
- **Function with Positional & Default Arguments**
- **Function with Keyword Arguments**
- **Function with Arbitrary Arguments (`*args`, `**kwargs`)**
- **Lambda Functions (Anonymous Functions)**
- **List Comprehension**
- **Dictionary Comprehension**

## 1. Function without Return Value

### Syntax
```python
def function_name(parameters):
    # statement(s) to execute without return
```

````python
# Function without return value
def greet_user(name):
    print(f"Hello, {name}! Welcome.")

result = greet_user("Alice")
print("Returned Value:", result)
````

### Explanation
1. A function without an explicit `return` executes its internal statements and finishes.
2. In Python, all functions without a specified return statement implicitly return `None`.

---
## 2. Function with Return Value

### Syntax
```python
def function_name(parameters):
    # statement(s)
    return value
```

````python
# Function with return value
def calculate_square(num):
    return num * num

output = calculate_square(6)
print("Square of 6:", output)
````

### Explanation
1. The `return` keyword sends the computed result back to the caller and terminates execution.
2. The returned output can be saved in a variable, reused in formulas, or passed to other functions.

---
## 3. Function with Positional & Default Arguments

### Syntax
```python
def function_name(required_arg, default_arg=default_value):
    # statement(s)
```

````python
# Function with positional and default arguments
def power(base, exponent=2):
    return base ** exponent

print("Default exponent (3^2):", power(3))
print("Custom exponent (3^3):", power(3, 3))
````

### Explanation
1. Positional arguments are mandatory and assigned sequentially according to order of input.
2. Default arguments provide fallback values when the caller omits optional arguments.

---
## 4. Function with Keyword Arguments

### Syntax
```python
def function_name(param1, param2):
    # statement(s)

# Calling by explicit parameter name
function_name(param2=val2, param1=val1)
```

````python
# Function called with keyword arguments
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

describe_pet(pet_name="Buddy", animal_type="dog")
````

### Explanation
1. Keyword arguments pass arguments using `parameter_name=value` pairs directly at call time.
2. This allows arguments to be specified in any order regardless of their defined position.

---
## 5. Function with Arbitrary Arguments (`*args` and `**kwargs`)

### Syntax
```python
def function_name(*args, **kwargs):
    # *args collects positional arguments into a tuple
    # **kwargs collects keyword arguments into a dictionary
```

````python
# Arbitrary arguments (*args and **kwargs)
def display_info(*skills, **details):
    print("Skills Tuple (*args):", skills)
    print("Details Dict (**kwargs):", details)

display_info("Python", "Django", name="Bob", experience=3)
````

### Explanation
1. `*args` bundles any number of excess positional arguments into a tuple.
2. `**kwargs` bundles any number of excess keyword arguments into a key-value dictionary.

---
## 6. Lambda Functions (Anonymous Functions)

### Syntax
```python
lambda_name = lambda parameters: expression
```

````python
# Lambda function
multiply = lambda a, b: a * b

print("Multiply (4 * 5):", multiply(4, 5))
````

### Explanation
1. A lambda function is an anonymous inline function defined without the `def` keyword.
2. It accepts any number of arguments but evaluates and returns only a single expression.

---
## 7. List Comprehension

### Syntax
```python
new_list = [expression for item in iterable if condition]
```

````python
# List comprehension (squares of even numbers from 1 to 6)
even_squares = [x ** 2 for x in range(1, 7) if x % 2 == 0]

print("Even Squares List:", even_squares)
````

### Explanation
1. List comprehension provides a concise one-line syntax to transform and filter an iterable into a list.
2. It combines mapping (`x ** 2`), iteration (`for x in ...`), and filtering (`if x % 2 == 0`) inside brackets.

---
## 8. Dictionary Comprehension

### Syntax
```python
new_dict = {key_expr: value_expr for item in iterable if condition}
```

````python
# Dictionary comprehension (numbers as keys and their cubes as values)
cubes_dict = {n: n ** 3 for n in range(1, 5)}

print("Cubes Dictionary:", cubes_dict)
````

### Explanation
1. Dictionary comprehension constructs a new dictionary dynamically from an iterable using curly braces.
2. Each cycle evaluates a `key: value` pair (`n: n ** 3`) to insert into the resulting mapping.

# Object-Oriented Programming (OOP) in Python

OOP is a programming paradigm based on the concept of 'objects' containing data (attributes) and code (methods).
Topics covered:
- **Classes & Objects**
- **CRUD Operations on Objects**
- **Inheritance**
- **Polymorphism**
- **Encapsulation**
- **Abstraction**

## 1. Classes & Objects

### Syntax
```python
class ClassName:
    def __init__(self, param1, param2):
        self.attribute1 = param1
        self.attribute2 = param2

    def method_name(self):
        # statement(s)

# Object instantiation
object_name = ClassName(val1, val2)
```

````python
# Class definition and Object instantiation
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Student: {self.name}, Age: {self.age}")

s1 = Student("John", 20)
s1.display()
````

### Explanation
1. A class serves as a blueprint defining state and behavior, while an object is an instance of that class.
2. The `__init__` constructor initializes instance attributes, and `self` references the current instance.

---
## 2. CRUD Operations on Objects

### Syntax
```python
# Create: obj.attr = val      (dynamically assign attribute)
# Read:   value = obj.attr    (access attribute)
# Update: obj.attr = new_val  (modify attribute value)
# Delete: del obj.attr        (delete attribute from instance)
```

````python
# CRUD Operations on Object Attributes
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

item = Product("Book", 15)  # Instantiation

# 1. Create (Add dynamic attribute)
item.in_stock = True

# 2. Read
print("Read:", item.name, item.price, item.in_stock)

# 3. Update
item.price = 18
print("Updated Price:", item.price)

# 4. Delete attribute
del item.in_stock
print("Has 'in_stock' after del?:", hasattr(item, "in_stock"))
````

### Explanation
1. Python objects support dynamic attribute creation, reading, and updating via standard dot-notation.
2. The `del` keyword completely deletes an attribute from an object, verified by `hasattr()`.

---
## 3. Inheritance

### Syntax
```python
class ParentClass:
    # parent attributes and methods

class ChildClass(ParentClass):
    # inherits parent and adds/overrides features
```

````python
# Inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks.")

d = Dog("Buddy")
d.speak()
````

### Explanation
1. Inheritance allows a child class (`Dog`) to derive attributes and methods from a parent class (`Animal`).
2. It promotes code reusability and allows derived classes to override methods with specific implementations.

---
## 4. Polymorphism

### Syntax
```python
class ClassA:
    def action(self): ...

class ClassB:
    def action(self): ...

# Same method name called across different object types
```

````python
# Polymorphism
class Cat:
    def sound(self):
        return "Meow"

class Cow:
    def sound(self):
        return "Moo"

animals = [Cat(), Cow()]
for a in animals:
    print("Animal sound:", a.sound())
````

### Explanation
1. Polymorphism enables distinct classes to share the same method name (`sound()`) with specialized behaviors.
2. A common caller loop executes each object's respective implementation without type-checking.

---
## 5. Encapsulation

### Syntax
```python
class ClassName:
    def __init__(self):
        self.__private_attr = value  # '__' prefix restricts direct external access
```

````python
# Encapsulation
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print("Balance via getter method:", acc.get_balance())
````

### Explanation
1. Encapsulation restricts direct modification of internal attributes by prefixing them with double underscores (`__`).
2. Safe access and updates are enforced exclusively through defined getter and setter methods.

---
## 6. Abstraction

### Syntax
```python
from abc import ABC, abstractmethod

class AbstractBase(ABC):
    @abstractmethod
    def mandatory_method(self):
        pass
```

````python
# Abstraction
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rect = Rectangle(4, 5)
print("Rectangle Area:", rect.area())
````

### Explanation
1. Abstraction defines common blueprint interfaces while hiding the internal implementation using `ABC`.
2. Subclasses must implement all methods decorated with `@abstractmethod` before they can be instantiated.
